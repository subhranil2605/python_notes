### augmented union operator (`|=`)

```python
config = {
    "color": "green",
    "width": 420,
    "height": 700,
    "font": "Courier"
}

user_config = {
    "path": "/home",
    "color": "black",
    "font": "Arial",
    "position": (200, 3000),
}
```

```python
config |= user_config
print(config)
```

```python
names = ["Rutuja", "Subhranil", "Raj"]
scores = [99, 85, 45]
```

## Function  - Return, **kwargs

```python
# Whenever we are not returning anything from a function, it will return None
def func_name(a, b):
    x = a ** b
    print(x)
```

```python
result = func_name(5, 6)
```

```python
print(result)
```

```python
def func_name(a, b) :
    x = a**b
    return x

result = func_name(4, 5)
print(result)
```
